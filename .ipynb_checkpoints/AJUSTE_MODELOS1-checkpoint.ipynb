## SARFIMA

### Pesos fraccionales

```python
def pesos_diferenciacion_fraccional(d, n):

    pesos = np.ones(n)

    for k in range(1, n):
        pesos[k] = -pesos[k - 1] * (d - k + 1) / k

    return pesos
```

### Diferenciación fraccional

```python
def diferenciacion_fraccional(serie, d):

    serie = np.asarray(serie, dtype=float)
    pesos = pesos_diferenciacion_fraccional(d=d, n=len(serie))
    serie_diff = np.convolve(serie, pesos, mode="full")[:len(serie)]

    return serie_diff
```

### Clase SARFIMA

```python
@dataclass
class ModeloSARFIMA:

    modelo: object
    d_fraccional: float
    historia: np.ndarray
    periodo_estacional: int

    def predict(self, n_periods):

        # Predicción sobre la serie diferenciada
        pred_diff = np.asarray(self.modelo.predict(n_periods=n_periods), dtype=float)

        historia = list(np.asarray(self.historia, dtype=float))
        n_train = len(historia)

        # Se calculan suficientes pesos también para las observaciones futuras
        pesos = pesos_diferenciacion_fraccional(d=self.d_fraccional, n=n_train + n_periods)

        predicciones = []

        # Se reconstruye la serie original
        for z_t in pred_diff:

            t = len(historia)
            acumulado = 0.0

            for k in range(1, t + 1):
                acumulado += (pesos[k] * historia[t - k])

            y_t = z_t - acumulado
            historia.append(y_t)
            predicciones.append(y_t)

        return np.asarray(predicciones)
```

### Ajuste de SARFIMA

```python
def ajustar_sarfima(y, m=7, d_grid=np.arange(0.05, 0.55, 0.1),
                    max_p=3, max_q=3, max_P=2, max_Q=2):

    mejor_modelo, mejor_aicc, mejor_d = None, np.inf, None

    # Rejilla de modelos
    for d_frac in d_grid:
        try:
            y_frac = diferenciacion_fraccional(y, d_frac)
            modelo = auto_arima(y_frac,

                # Parte ARIMA
                start_p=0, start_q=0, max_p=max_p, max_q=max_q, d=0,

                # Parte estacional
                seasonal=m > 1, m=m, start_P=0, start_Q=0, max_P=max_P, max_Q=max_Q, D=None if m > 1 else 0, max_D=1,

                # Selección automática
                information_criterion="aicc", stepwise=True, error_action="ignore", suppress_warnings=True)

            if np.isfinite(modelo.aicc()) and modelo.aicc() < mejor_aicc:
                mejor_modelo = modelo
                mejor_aicc = modelo.aicc()
                mejor_d = d_frac
        except:
            continue

    # Modelo por default
    if mejor_modelo is None:
        try:
            d_frac = 0.1
            y_frac = diferenciacion_fraccional(y, d_frac)
            mejor_modelo = auto_arima(y_frac, d=0, seasonal=False, max_p=2, max_q=2, stepwise=True, error_action="ignore", suppress_warnings=True)
            mejor_d = d_frac
        except:
            return None


    return ModeloSARFIMA(modelo=mejor_modelo, d_fraccional=mejor_d, historia=np.asarray(y), periodo_estacional=m)
```

## SARIMA

```python
def ajustar_sarima(y, m=7, max_p=3, max_q=3, max_P=2, max_Q=2):

    modelo = auto_arima(y,

        # Parte ARIMA
        start_p=0, start_q=0, max_p=max_p, max_q=max_q, d=None, max_d=2,

        # Parte estacional
        seasonal=(m > 1), m=m, start_P=0, start_Q=0, max_P=max_P, max_Q=max_Q, D=None, max_D=1,

        # Selección automática
        information_criterion="aicc", test="kpss", seasonal_test="ocsb", stepwise=True,
        suppress_warnings=True, error_action="ignore", with_intercept="auto", maxiter=100, trace=False
    )

    return modelo
```

## GARCH

```python
def ajustar_garch(y, max_ar=4, max_p=3, max_q=3, distribuciones=("normal", "t")):
    mejor_modelo, mejor_bic = None, np.inf

    # Rejilla de modelos
    for ar in range(max_ar + 1):
        for p in range(1, max_p + 1):
            for q in range(1, max_q + 1):
                for dist in distribuciones:
                    try:
                        modelo = arch_model(y, mean="AR" if ar > 0 else "Constant", lags=ar if ar > 0 else None,
                                            vol="GARCH", p=p, q=q, dist=dist)
                        ajuste = modelo.fit(disp="off", show_warning=False)

                        if np.isfinite(ajuste.bic) and ajuste.bic < mejor_bic:
                            mejor_modelo = ajuste
                            mejor_bic = ajuste.bic
                    except:
                        continue

    # Modelo por defaut
    if mejor_modelo is None:
        try:
            mejor_modelo = arch_model(y, mean="Constant", vol="GARCH", p=1, q=1).fit(disp="off", show_warning=False)
        except:
            pass

    return mejor_modelo
```

## Ajuste de modelos

### Métrica de comparación

```python
def calcular_error(y_real, y_pred):

    y_real = np.asarray(y_real, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    return mean_absolute_error(y_real, y_pred)
```

```python
def seleccionar_mejor_modelo(df, df_test, m=7):
    mejores_modelos, resultados = [], []

    for columna in df.columns:
        y_train = df[columna].dropna().values
        y_test = df_test[columna].dropna().values
        horizonte = len(y_test)
        candidatos = {}

        # GARCH
        modelo = ajustar_garch(y_train)
        if modelo is not None:
            try:
                pred = modelo.forecast(horizon=horizonte, reindex=False).mean.iloc[-1].values
                candidatos["GARCH"] = (modelo, np.sqrt(mean_squared_error(y_test, pred)))
            except:
                pass

        # SARIMA
        modelo = ajustar_sarima(y_train, m=m)
        if modelo is not None:
            try:
                pred = modelo.predict(n_periods=horizonte)
                candidatos["SARIMA"] = (modelo, np.sqrt(mean_squared_error(y_test, pred)))
            except:
                pass

        # SARFIMA
        modelo = ajustar_sarfima(y_train, m=m)
        if modelo is not None:
            try:
                pred = modelo.predict(n_periods=horizonte)
                candidatos["SARFIMA"] = (modelo, np.sqrt(mean_squared_error(y_test, pred)))
            except:
                pass

        # Elección de candidatos
        if candidatos:
            ganador = min(candidatos, key=lambda x: candidatos[x][1])
            modelo, mae = candidatos[ganador]
            mejores_modelos.append(modelo)
            resultados.append({"serie": columna, "modelo": ganador, "MAE": mae})

    return mejores_modelos, pd.DataFrame(resultados)
```

### Diagnóstico y recuperación de series con `error = -1`

Esta sección vuelve a procesar únicamente las series cuyo error sigue marcado como `-1`. A diferencia del intento original:

- convierte cada serie a valores numéricos y trata los infinitos como ausentes;
- conserva la separación temporal 90/10 antes de cualquier relleno;
- imputa solamente el conjunto de entrenamiento, sin utilizar datos futuros;
- evalúa las predicciones solo en posiciones observadas del conjunto de prueba;
- prueba primero la periodicidad solicitada y luego una versión no estacional;
- registra el motivo concreto por el que falla cada candidato;
- usa `NAIVE_LAST` como último recurso reproducible cuando los tres modelos complejos fallan;
- conserva `-1` únicamente cuando ni siquiera se puede construir o evaluar ese recurso.

La columna histórica `error` se conserva por compatibilidad, pero su valor corresponde a **RMSE**, que es la métrica calculada por el código original.


```python
from dataclasses import dataclass
from pathlib import Path
import re


@dataclass
class ModeloNaiveUltimoValor:
    """Pronostica el último valor observado del conjunto de entrenamiento."""
    ultimo_valor: float

    def predict(self, n_periods):
        return np.repeat(float(self.ultimo_valor), int(n_periods))


def _nombre_archivo_seguro(valor):
    texto = re.sub(r"[^A-Za-z0-9._-]+", "_", str(valor)).strip("._")
    return texto or "serie"


def _rmse_valido(y_test_completo, pred):
    """Evalúa posiciones observadas y verifica longitud y valores finitos."""
    y_test_completo = np.asarray(y_test_completo, dtype=float)
    pred = np.asarray(pred, dtype=float).reshape(-1)

    if len(pred) != len(y_test_completo):
        raise ValueError(
            f"longitud de predicción {len(pred)} distinta del horizonte "
            f"{len(y_test_completo)}"
        )

    mascara = np.isfinite(y_test_completo)
    if not mascara.any():
        raise ValueError("el conjunto de prueba no contiene observaciones válidas")
    if not np.isfinite(pred[mascara]).all():
        raise ValueError("la predicción contiene NaN o infinito")

    return float(np.sqrt(mean_squared_error(
        y_test_completo[mascara], pred[mascara]
    )))


def reprocesar_serie(serie, m=7, proporcion_train=0.9):
    """Ajusta candidatos y devuelve modelo, resultado y diagnóstico."""
    y = pd.to_numeric(serie, errors="coerce").replace([np.inf, -np.inf], np.nan)
    n_total = len(y)

    if n_total < 3:
        return None, None, {
            "estado": "fallido",
            "detalle": "menos de tres registros en la serie",
            "n_total": n_total,
            "n_train": 0,
            "n_test_observado": 0,
        }

    corte = min(max(int(n_total * proporcion_train), 1), n_total - 1)
    train_raw = y.iloc[:corte]
    test_raw = y.iloc[corte:]

    # Se usan únicamente valores del entrenamiento para completar el entrenamiento.
    train = train_raw.interpolate(method="linear", limit_direction="both")
    train = train.ffill().bfill()
    y_train = train.to_numpy(dtype=float)
    y_test = test_raw.to_numpy(dtype=float)
    n_test_observado = int(np.isfinite(y_test).sum())

    diagnostico_base = {
        "n_total": n_total,
        "n_train": int(np.isfinite(y_train).sum()),
        "n_test_observado": n_test_observado,
    }

    if not np.isfinite(y_train).any():
        return None, None, {
            **diagnostico_base,
            "estado": "fallido",
            "detalle": "el entrenamiento no contiene observaciones válidas",
        }
    if n_test_observado == 0:
        return None, None, {
            **diagnostico_base,
            "estado": "fallido",
            "detalle": "la prueba no contiene observaciones válidas para evaluar",
        }

    horizonte = len(y_test)
    candidatos = []
    errores = []

    # Una periodicidad semanal necesita suficiente historia. Siempre se prueba m=1.
    periodos = []
    if m > 1 and len(y_train) >= 3 * m:
        periodos.append(m)
    periodos.append(1)
    periodos = list(dict.fromkeys(periodos))

    for periodo in periodos:
        configuracion = f"m={periodo}"

        try:
            modelo = ajustar_sarima(y_train, m=periodo)
            pred = modelo.predict(n_periods=horizonte)
            candidatos.append(("SARIMA", modelo, _rmse_valido(y_test, pred), periodo))
        except Exception as exc:
            errores.append(f"SARIMA({configuracion}): {type(exc).__name__}: {exc}")

        try:
            modelo = ajustar_sarfima(y_train, m=periodo)
            if modelo is None:
                raise RuntimeError("el ajuste devolvió None")
            pred = modelo.predict(n_periods=horizonte)
            candidatos.append(("SARFIMA", modelo, _rmse_valido(y_test, pred), periodo))
        except Exception as exc:
            errores.append(f"SARFIMA({configuracion}): {type(exc).__name__}: {exc}")

    # GARCH no utiliza m, por lo que se intenta una sola vez.
    try:
        modelo = ajustar_garch(y_train)
        if modelo is None:
            raise RuntimeError("el ajuste devolvió None")
        pred = modelo.forecast(horizon=horizonte, reindex=False).mean.iloc[-1].values
        candidatos.append(("GARCH", modelo, _rmse_valido(y_test, pred), None))
    except Exception as exc:
        errores.append(f"GARCH: {type(exc).__name__}: {exc}")

    if candidatos:
        nombre, modelo, rmse, periodo = min(candidatos, key=lambda x: x[2])
        return modelo, {
            "modelo": nombre,
            "RMSE": rmse,
            "m_usado": periodo,
        }, {
            **diagnostico_base,
            "estado": "recuperado_modelo",
            "detalle": " | ".join(errores) if errores else "sin errores de candidatos",
        }

    # Último recurso: ofrece una referencia evaluable y queda identificado explícitamente.
    try:
        modelo = ModeloNaiveUltimoValor(float(y_train[-1]))
        pred = modelo.predict(horizonte)
        rmse = _rmse_valido(y_test, pred)
        return modelo, {
            "modelo": "NAIVE_LAST",
            "RMSE": rmse,
            "m_usado": None,
        }, {
            **diagnostico_base,
            "estado": "recuperado_naive",
            "detalle": " | ".join(errores) or "fallaron los modelos complejos",
        }
    except Exception as exc:
        errores.append(f"NAIVE_LAST: {type(exc).__name__}: {exc}")
        return None, None, {
            **diagnostico_base,
            "estado": "fallido",
            "detalle": " | ".join(errores),
        }


# Recalcular todos los marcadores -1 y actualizar por clave de forma verificable.
import hashlib
import unicodedata


def _normalizar_nombre(valor):
    texto = unicodedata.normalize("NFKC", str(valor)).strip().casefold()
    return " ".join(texto.split())


def _resolver_columna(indice, fila, columnas_df):
    """Busca el nombre de serie en el índice y en columnas descriptivas."""
    candidatos = [indice]
    for campo in ("serie", "variable", "nombre_columna", "columna"):
        if campo in fila.index and pd.notna(fila[campo]):
            candidatos.append(fila[campo])

    mapa = {}
    for columna in columnas_df:
        mapa.setdefault(_normalizar_nombre(columna), []).append(columna)

    for candidato in candidatos:
        coincidencias = mapa.get(_normalizar_nombre(candidato), [])
        if len(coincidencias) == 1:
            return coincidencias[0], None
        if len(coincidencias) > 1:
            return None, f"nombre ambiguo: {candidato!r}"

    return None, f"no existe una columna correspondiente a {indice!r} en df"


def _ruta_modelo_corta(modelo_nombre, serie_nombre):
    """Evita fallos de Windows por nombres o rutas demasiado largas."""
    identificador = hashlib.sha1(
        str(serie_nombre).encode("utf-8")
    ).hexdigest()[:12]
    return Path("modelos") / (
        f"{_nombre_archivo_seguro(modelo_nombre)[:24]}_{identificador}.joblib"
    )


if "error" not in mejores_modelos_df.columns:
    raise KeyError("mejores_modelos_df no contiene la columna 'error'")

# Admite -1 numérico y '-1' almacenado como texto.
error_numerico = pd.to_numeric(mejores_modelos_df["error"], errors="coerce")
indices_pendientes = list(dict.fromkeys(
    mejores_modelos_df.index[error_numerico.eq(-1)].tolist()
))

print(f"Series marcadas con error = -1: {len(indices_pendientes)}")

# Las columnas permiten conservar el resultado aun cuando el modelo no sea evaluable.
for columna_estado in ("estado_recalculo", "detalle_recalculo"):
    if columna_estado not in mejores_modelos_df.columns:
        mejores_modelos_df[columna_estado] = pd.NA

Path("modelos").mkdir(parents=True, exist_ok=True)
registros_diagnostico = []
cache_por_columna = {}

for numero, indice in enumerate(indices_pendientes, start=1):
    fila = mejores_modelos_df.loc[indice]
    if isinstance(fila, pd.DataFrame):
        fila = fila.iloc[0]

    columna_df, error_nombre = _resolver_columna(indice, fila, df.columns)
    print(f"[{numero}/{len(indices_pendientes)}] {indice}", end="")

    if error_nombre is not None:
        modelo = resultado = None
        diagnostico = {
            "estado": "no_evaluable",
            "detalle": error_nombre,
            "n_total": 0,
            "n_train": 0,
            "n_test_observado": 0,
        }
    elif columna_df in cache_por_columna:
        modelo, resultado, diagnostico = cache_por_columna[columna_df]
        diagnostico = dict(diagnostico)
        diagnostico["detalle"] = (
            diagnostico.get("detalle", "") + " | resultado reutilizado"
        ).strip(" |")
    else:
        modelo, resultado, diagnostico = reprocesar_serie(
            df[columna_df], m=7, proporcion_train=0.9
        )
        cache_por_columna[columna_df] = (modelo, resultado, dict(diagnostico))

    registro = {
        "indice_resultados": indice,
        "serie_df": columna_df,
        **diagnostico,
        "modelo": None,
        "RMSE": np.nan,
        "m_usado": None,
        "archivo_modelo": None,
    }

    if modelo is not None and resultado is not None:
        ganador = resultado["modelo"]
        rmse = float(resultado["RMSE"])

        # Primero se actualiza el cálculo. Un fallo de persistencia no invalida el RMSE.
        mejores_modelos_df.loc[indice, "mejor_modelo"] = ganador
        mejores_modelos_df.loc[indice, "error"] = rmse
        mejores_modelos_df.loc[indice, "estado_recalculo"] = diagnostico["estado"]
        mejores_modelos_df.loc[indice, "detalle_recalculo"] = diagnostico["detalle"]

        ruta = _ruta_modelo_corta(ganador, columna_df)
        try:
            joblib.dump(modelo, ruta, compress=3)
            registro["archivo_modelo"] = str(ruta)
        except Exception as exc:
            registro["estado"] = "recalculado_sin_archivo"
            registro["detalle"] = (
                f"{registro['detalle']} | guardado: "
                f"{type(exc).__name__}: {exc}"
            ).strip(" |")
            mejores_modelos_df.loc[indice, "estado_recalculo"] = registro["estado"]
            mejores_modelos_df.loc[indice, "detalle_recalculo"] = registro["detalle"]

        registro.update({
            "modelo": ganador,
            "RMSE": rmse,
            "m_usado": resultado["m_usado"],
        })
        print(f" -> {ganador}, RMSE={rmse:.6g}")
    else:
        # -1 es un marcador de pendiente, no una métrica válida. Si no se puede
        # evaluar, se usa NaN y se conserva la causa explícita.
        mejores_modelos_df.loc[indice, "mejor_modelo"] = "NO_EVALUABLE"
        mejores_modelos_df.loc[indice, "error"] = np.nan
        mejores_modelos_df.loc[indice, "estado_recalculo"] = diagnostico["estado"]
        mejores_modelos_df.loc[indice, "detalle_recalculo"] = diagnostico["detalle"]
        print(f" -> NO_EVALUABLE: {diagnostico['detalle']}")

    registros_diagnostico.append(registro)

    # Punto de recuperación: evita perder todos los resultados si el proceso se interrumpe.
    mejores_modelos_df.to_csv("mejores_modelos.csv")
    pd.DataFrame(registros_diagnostico).to_csv(
        "reporte_reprocesamiento.csv", index=False
    )

reporte_reprocesamiento = pd.DataFrame(registros_diagnostico)
error_final = pd.to_numeric(mejores_modelos_df["error"], errors="coerce")
pendientes_finales = mejores_modelos_df.loc[error_final.eq(-1)]

print("\nResumen del recálculo:")
if not reporte_reprocesamiento.empty:
    display(reporte_reprocesamiento)

print(f"Series recuperadas o clasificadas: {len(registros_diagnostico)}")
print(f"Series que aún conservan el marcador -1: {len(pendientes_finales)}")

if not pendientes_finales.empty:
    display(pendientes_finales)
    raise RuntimeError(
        "Persisten valores -1; revise duplicados de índice o modificaciones "
        "concurrentes en mejores_modelos_df."
    )
```
