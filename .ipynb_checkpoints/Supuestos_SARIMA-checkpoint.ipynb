En este archivo se evalúan los supuestos para los modelos SARIMA correspondientes.

```python
#LIBRERIAS IMPORTADAS
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch
import statsmodels.api as sm
from scipy.stats import jarque_bera
```

```python
#Funciones a utilizar
def grafico_residuos(modelo, nombre):

    residuos = modelo.resid.dropna()

    plt.figure(figsize=(12, 4))

    plt.plot(residuos)

    plt.axhline(
        0,
        linestyle="--"
    )

    plt.title(
        f"Residuos - {nombre}"
    )

    plt.xlabel("Tiempo")
    plt.ylabel("Residuo")

    plt.show()


def grafico_acf_residuos(modelo, nombre, lags=36):

    residuos = modelo.resid.dropna()

    plot_acf(
        residuos,
        lags=lags
    )

    plt.title(
        f"ACF de residuos - {nombre}"
    )

    plt.show()
    
def histograma_residuos(modelo, nombre):

    residuos = modelo.resid.dropna()

    plt.figure(figsize=(8, 5))

    plt.hist(
        residuos,
        bins=30,
        density=True
    )

    plt.title(
        f"Distribución de residuos - {nombre}"
    )

    plt.xlabel("Residuo")

    plt.show()

def qq_residuos(modelo, nombre):

    residuos = modelo.resid.dropna()

    sm.qqplot(
        residuos,
        line="45"
    )

    plt.title(
        f"Q-Q plot - {nombre}"
    )

    plt.show()
```
