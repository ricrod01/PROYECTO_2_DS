```python
from modelo_SARFIMA import ModeloSARFIMA
import joblib
import os
from pathlib import Path
```

```python
carpeta = Path("modelos")

for ruta in carpeta.glob("*.joblib"):
    peso_antes = ruta.stat().st_size / 1024**2

    try:
        modelo = joblib.load(ruta)
        temporal = str(ruta) + ".tmp"

        joblib.dump(modelo, temporal, compress=3)
        os.replace(temporal, ruta)

        peso_despues = ruta.stat().st_size / 1024**2
        print(f"{ruta.name}: {peso_antes:.2f} MB -> {peso_despues:.2f} MB")
    except Exception as e:
        print(f"Error en {ruta.name}: {e}")
```
